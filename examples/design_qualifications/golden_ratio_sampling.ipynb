# Generalized golden ratio sampling
Generalized golden ratio sampling is a multi-dimensional, quasi-random, low-discrepancy sequence suggestor based on the generalized golden ratio. For a lot more details, see https://extremelearning.com.au/unreasonable-effectiveness-of-quasirandom-sequences/

Here, we show how easy it is to use with XPyrimentor.


```python
import matplotlib.pyplot as plt
from ProcessOptimizer.XpyriMentor import XpyriMentor

# Create a sampler that uses the generalized golden ratio sampling
xpyrimentor = XpyriMentor(space = [(0.0, 1.0), (0.0, 1.0)], suggestor={"suggestor_name": "GoldenRatio"})
# Ask for 120 points in the space defined by the suggestor
points = xpyrimentor.ask(120)
# Plot the points in the space
plt.plot([point[0] for point in points], [point[1] for point in points], 'ro')
```

We can also continue the sequence later on. We just have to tell XPyriMentor about the points. The generalized golden ratio sampling will then continue the pattern from before. Notice that it "fills out" the "holes" left by the red points.


Note that we need to give XPyriMentor a value for the objetive in the point even though the generalized golden ratio sampling doesn't use it. This
is to allow for later use of data dependent sampling, e.g. Bayesian optimization.

```python
# "telling" the XpyriMentor about value in the points suggested.
for point in points:
    xpyrimentor.tell(point, [0.0])  # Assuming a dummy objective value of 0.0
# Get 30 new points in the space
new_points = xpyrimentor.ask(30)
# Plot both the original and new points
plt.plot([point[0] for point in points], [point[1] for point in points], 'ro')
plt.plot([point[0] for point in new_points], [point[1] for point in new_points], 'bo')
```
