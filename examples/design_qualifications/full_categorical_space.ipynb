# Fully categorical space

If you only want to optimize categorical parameters, the default regressor
(Gaussian process regressor) does not work, and you will get an error.

```python
from ProcessOptimizer import Optimizer

space = [
    ["dog", "cat", "mouse"],
    ["swim", "jump", "run"],
]

gp_opt = Optimizer(space)
```

Instead, it is recommended to use a random forest regressor.

```python
rf_opt = Optimizer(space, base_estimator="RF", n_initial_points=3)
rf_opt.tell(["dog", "swim"], -1)
rf_opt.tell(["dog", "jump"], -2)
rf_opt.tell(["mouse", "run"], -2)

rf_opt.ask()
```
